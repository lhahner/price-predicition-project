# Feature Engineering

#### Comments to sentiment scala -1 to 1

```python
import pandas as pd
from textblob import TextBlob
```

```python
clean_listing_df = pd.read_csv('./processed-data/clean-data.csv')
#create small df with id and comment
feature_df = clean_listing_df[["comments","listing_id"]] 
#add sentiment score for comment
feature_df['comment_score'] = feature_df['comments'].apply(lambda x: TextBlob(x).sentiment.polarity if pd.notnull(x) else 0)
#drop original comment
feature_df = feature_df.drop(columns=['comments'])
#avg sentiment per id
feature_df = feature_df.groupby('listing_id', as_index=False)['comment_score'].mean()
```

```python
feature_df
```

```python
listings_df = pd.read_csv("./processed-data/clean-listings.csv")

# 2. Füge den durchschnittlichen Sentiment-Wert hinzu (merge)
listings_df = listings_df.merge(feature_df, on="listing_id", how="left")

# 3. Speichere das Ergebnis wieder als neue CSV
listings_df.to_csv("./processed-data/clean-data-with-sentiment.csv", index=False)
```
